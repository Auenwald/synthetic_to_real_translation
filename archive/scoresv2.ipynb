```python
import json
import numpy as np

def extract_mean_ious(json_path):
    # JSON laden
    with open(json_path, "r") as f:
        data = json.load(f)

    # Ergebnis-Dict vorbereiten
    result = {}
    for dataset, epochs in data.items():
        result[dataset] = {}
        for epoch, metrics in epochs.items():
            result[dataset][int(epoch)] = metrics["mean_iou"]
    return result


def print_mious(mean_ious, skip_first_n=5):
    synthia_mious = []
    bdd_mious = []
    cityscapes_mious = []

    for dataset, epochs in mean_ious.items():
        for epoch, value in epochs.items():
            if dataset.lower() == 'synthia':
                synthia_mious.append((epoch, value))
            elif dataset.lower() == 'cityscapes':
                cityscapes_mious.append((epoch, value))
            elif dataset.lower() == 'bdd':
                bdd_mious.append((epoch, value))

    # Nach Epoch sortieren, falls JSON unsortiert ist
    synthia_mious.sort(key=lambda x: x[0])
    cityscapes_mious.sort(key=lambda x: x[0])
    bdd_mious.sort(key=lambda x: x[0])

    # Skip erste N Epochen
    synthia_epochs, synthia_vals = zip(*synthia_mious)
    cityscapes_epochs, cityscapes_vals = zip(*cityscapes_mious)
    bdd_epochs, bdd_vals = zip(*bdd_mious)

    synthia_mious = np.array(synthia_vals[skip_first_n:])
    cityscapes_mious = np.array(cityscapes_vals[skip_first_n:])
    bdd_mious = np.array(bdd_vals[skip_first_n:])

    # Best checkpoint anhand Synthia
    idx_synthia_epoch = np.argmax(synthia_mious)

    cityscapes_cp_miou, bdd_cp_miou = cityscapes_mious[idx_synthia_epoch], bdd_mious[idx_synthia_epoch]
    cityscapes_m_miou, bdd_m_miou = np.mean(cityscapes_mious), np.mean(bdd_mious)
    cityscapes_max_miou, bdd_max_miou = np.max(cityscapes_mious), np.max(bdd_mious)

    # Epoch-Bereich für Transparenz
    start_epoch = synthia_epochs[skip_first_n]
    end_epoch = synthia_epochs[-1]

    print(f'Mean calculated over epochs {start_epoch}–{end_epoch}')
    print(f'Cityscapes mean IoU: {cityscapes_m_miou:.3f}, CP mIoU: {cityscapes_cp_miou:.3f}, max mIoU {cityscapes_max_miou:.3f}')
    print(f'BDD mean IoU: {bdd_m_miou:.3f}, CP mIoU: {bdd_cp_miou:.3f}, max mIoU {bdd_max_miou:.3f}')
```

```python
path = "./logs/synthia_to_cs_and_bdd_segformerb5_lr1e5_lr1e4_no_crossattention_baseline.json"
mean_ious = extract_mean_ious(path)
print_mious(mean_ious)
```
