```python
import json
import numpy as np

def extract_mean_ious(json_path):
    # JSON laden
    with open(json_path, "r") as f:
        data = json.load(f)

    # Ergebnis-Dict vorbereiten
    result = {}
    for dataset, epochs in data.items():
        result[dataset] = {}
        for epoch, metrics in epochs.items():
            result[dataset][int(epoch)] = metrics["mean_iou"]
    return result

'''
if __name__ == "__main__":
    path = "./logs/synthia_to_cs_and_bdd_lr1e5_and_lr1e4_crossattention_rgb_and_lab_no_amp.json"
    mean_ious = extract_mean_ious(path)

    # Ausgabe
    for dataset, epochs in mean_ious.items():
        print(f"\n📊 {dataset.capitalize()}")
        for epoch in sorted(epochs.keys()):
            print(f"  Epoch {epoch:3d}: {epochs[epoch]:.3f}")
'''
```

```python
path = "./logs/synthia_to_cs_and_bdd_segformerb5_lr1e5_lr1e4_no_crossattention_baseline.json"
mean_ious = extract_mean_ious(path)

synthia_mious = []
bdd_mious = []
cityscapes_mious = []

for dataset, epochs in mean_ious.items():
    for epoch in epochs.keys():
        if dataset.lower() == 'synthia':
            synthia_mious.append(epochs[epoch])
        elif dataset.lower() == 'cityscapes':
            cityscapes_mious.append(epochs[epoch])
        elif dataset.lower() == 'bdd':
            bdd_mious.append(epochs[epoch])


synthia_mious = np.array(synthia_mious)
bdd_mious = np.array(bdd_mious)
cityscapes_mious = np.array(cityscapes_mious)

idx_synthia_epoch = np.argmax(synthia_mious)

cityscapes_cp_miou, bdd_cp_miou = cityscapes_mious[idx_synthia_epoch], bdd_mious[idx_synthia_epoch]
cityscapes_m_miou, bdd_m_miou = np.mean(cityscapes_mious), np.mean(bdd_mious)
cityscapes_max_miou, bdd_max_miou = np.max(cityscapes_mious), np.max(bdd_mious)
cityscapes_var_miou, bdd_var_miou = np.var(cityscapes_mious), np.var(bdd_mious)

print(f'Cityscapes mean mean IoU: {cityscapes_m_miou:.3f}, CP mIoU: {cityscapes_cp_miou:.3f}, max mIoU {cityscapes_max_miou:.3f}')
print(f'BDD mean mean IoU: {bdd_m_miou:.3f}, CP mIoU: {bdd_cp_miou:.3f}, max mIoU {bdd_max_miou:.3f}')
```
